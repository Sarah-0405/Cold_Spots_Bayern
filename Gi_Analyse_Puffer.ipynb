```python
import os

# Define the path to the "Puffergebiete" folder in Google Drive
drive_folder_path = "/content/drive/MyDrive/Cold Spots Bayern/Puffergebiete"

# Initialize dictionaries to store file paths categorized by city and time period
gpkg_files_2024 = {}
gpkg_files_2019_2024 = {}

# Check if the folder exists
if os.path.exists(drive_folder_path):
    # List all files in the folder
    for filename in os.listdir(drive_folder_path):
        # Check if the file is a GeoPackage file
        if filename.endswith(".gpkg"):
            file_path = os.path.join(drive_folder_path, filename)

            # Extract city name and determine time period from filename
            # Assuming the filename format is "joined_350mpuffer_um_Gitterzelle_mit_lst_[cityname].gpkg" or "joined_350mpuffer_um_Gitterzelle_mit_lst_2019_2024[cityname].gpkg"
            parts = filename.replace(".gpkg", "").split("_")
            if "2019" in parts:
                # This is a 2019-2024 file
                # The city name is the last part of the filename
                city_name = parts[-1].replace("2024", "") # Remove 2024 if present at the end
                gpkg_files_2019_2024[city_name] = file_path
            else:
                # This is a 2024 file
                # The city name is the last part of the filename
                city_name = parts[-1]
                gpkg_files_2024[city_name] = file_path

    # Print the categorized files
    print("GeoPackage files for 2024:")
    for city, path in gpkg_files_2024.items():
        print(f"  {city}: {path}")

    print("\nGeoPackage files for 2019-2024:")
    for city, path in gpkg_files_2019_2024.items():
        print(f"  {city}: {path}")

else:
    print(f"The folder '{drive_folder_path}' was not found. Please check the path.")
```

```python
import esda
from esda.getisord import G_Local
import gc
import libpysal
import libpysal.io
import pandas as pd
import os
import geopandas as gpd

# Define the directory to save the G_Local results in Google Drive
glocal_results_save_dir = "/content/drive/MyDrive/Cold Spots Bayern/Pufferzonen_gi_results"

# Create the directory in Google Drive if it doesn't exist
if not os.path.exists(glocal_results_save_dir):
    os.makedirs(glocal_results_save_dir)
    print(f"Created Google Drive directory for G_Local results: {glocal_results_save_dir}")
else:
    print(f"Google Drive directory for G_Local results already exists: {glocal_results_save_dir}")

# Define the path to the weights file in Google Drive
weights_file_path = "/content/drive/MyDrive/Cold Spots Bayern/Pufferzonen_KNN_Weights"

# Define the cities to process
cities_to_process = ["Aschaffenburg", "Würzburg"]


# --- Loading weights and performing G_Local analysis for selected cities ---

print("\n--- Step 2: Loading weights and performing G_Local analysis for selected cities ---")


# Function to calculate G_Local using a loaded weights matrix
def calculate_g_local_with_weights(gdf, wq):
    """
    Calculates Getis-Ord G_Local statistic using a pre-calculated weights matrix.

    Args:
        gdf (GeoDataFrame): Input GeoDataFrame with 'avg_summer_LST_Celsius' column.
        wq (libpysal.weights.W): Pre-calculated spatial weights matrix.

    Returns:
        tuple: A tuple containing the calculated G_Local values (as a pandas Series) and the p-values (as a pandas Series).
               Returns (None, None) if calculation fails.
    """
    try:
        # Ensure the GeoDataFrame's index matches the weights matrix index
        if not gdf.index.equals(pd.Int64Index(wq.id_order)):
            print("  Warning: GeoDataFrame index does not match weights matrix index. Attempting to reindex GeoDataFrame.")
            # Create a mapping from the weights matrix ID order to the GeoDataFrame index
            id_to_index = {id: index for index, id in enumerate(gdf.index)}
            # Reorder the weights matrix ID order to match the GeoDataFrame index
            wq.id_order = gdf.index.tolist()
            # Ensure weights are aligned with the re-indexed gdf
            wq.neighbors = {new_id: [wq.id_order[old_id_idx] for old_id_idx in neighbors]
                            for new_id, neighbors in zip(wq.id_order, wq.neighbors.values())}
            wq.weights = {new_id: weights for new_id, weights in zip(wq.id_order, wq.weights.values())}


        # Calculate Getis-Ord G_Local statistic
        g_local = G_Local(gdf['avg_summer_LST_Celsius'], wq)

        # Return G_Local values and p-values
        return pd.Series(g_local.Gs, index=gdf.index), pd.Series(g_local.p_sim, index=gdf.index)

    except Exception as e:
        print(f"  Error calculating G_Local: {e}")
        return None, None

# Load the weights matrix once
try:
    print(f"  Loading weights from: {weights_file_path}")
    wq = libpysal.io.open(weights_file_path, mode='r').read()
    print("  Weights loaded successfully.")
except Exception as e:
    print(f"  Error loading weights file: {e}")
    wq = None # Set wq to None if loading fails


if wq is not None:
    # Process 2019-2024 files for selected cities
    print("\n  Processing 2019-2024 data for selected cities:")
    for city_name, file_path in gpkg_files_2019_2024.items():
        if city_name in cities_to_process:
            print(f"  Processing {city_name} (2019-2024)")

            # Define the Google Drive save path for the current city and time period
            google_drive_save_path = os.path.join(glocal_results_save_dir, f"glocal_results_{city_name.replace(' ', '_')}_2019_2024.gpkg")

            # Check if the file already exists in Google Drive
            if os.path.exists(google_drive_save_path):
                print(f"  Results for {city_name} (2019-2024) already exist at '{google_drive_save_path}'. Skipping processing.")
                continue

            try:
                # Load the GeoPackage file
                gdf = gpd.read_file(file_path)
                print(f"  Loaded data for {city_name} (2019-2024).")

                # Ensure the GeoDataFrame has a valid projection for accurate G_Local calculation
                if gdf.crs is None or gdf.crs.is_geographic:
                     print("  Warning: GeoDataFrame has no CRS or is geographic. Reprojecting to EPSG:3035 for accurate G_Local calculation with loaded weights.")
                     gdf_projected_for_g = gdf.to_crs(epsg=3035) # Use a projected CRS
                else:
                     gdf_projected_for_g = gdf


                # Perform G_Local calculations using the function
                g_local_values, g_local_pvalues = calculate_g_local_with_weights(gdf_projected_for_g, wq)

                # Store the results
                if g_local_values is not None and g_local_pvalues is not None:
                     # Add the G_Local values and p-values back to the original gdf (non-projected)
                     gdf['g_local_2019_2024'] = g_local_values
                     gdf['g_local_pvalue_2019_2024'] = g_local_pvalues

                     # Filter for cold spots: p-value < 0.01 and G_Local value < 0
                     # Keep all columns for saving, not just the limited set in the previous version
                     cold_spots_gdf = gdf[(gdf['g_local_pvalue_2019_2024'] < 0.01) & (gdf['g_local_2019_2024'] < 0)].copy()

                     # Save the results to Google Drive as GeoPackage
                     if not cold_spots_gdf.empty:
                         try:
                             cold_spots_gdf.to_file(google_drive_save_path, driver='GPKG')
                             print(f"  Significant cold spots for {city_name} (2019-2024) saved to Google Drive: {google_drive_save_path}")
                         except Exception as save_e:
                             print(f"  Error saving results for {city_name} (2019-2024) to {google_drive_save_path}: {save_e}")

                     else:
                         print(f"  No significant cold spots found for {city_name} (2019-2024) with p < 0.01 and G_Local < 0. No file saved.")


            except Exception as e:
                print(f"  Error processing file {file_path}: {e}")
            finally:
                # Explicitly delete the GeoDataFrame
                if 'gdf' in locals():
                    del gdf
                if 'gdf_projected_for_g' in locals():
                    del gdf_projected_for_g
                if 'cold_spots_gdf' in locals():
                     del cold_spots_gdf
                gc.collect()
                print(f"  Finished G_Local processing for {city_name} (2019-2024). Memory freed.")


    # Process 2024 files for selected cities
    print("\n  Processing 2024 data for selected cities:")
    for city_name, file_path in gpkg_files_2024.items():
        if city_name in cities_to_process:
            print(f"  Processing {city_name} (2024)")

            # Define the Google Drive save path for the current city and time period
            google_drive_save_path = os.path.join(glocal_results_save_dir, f"glocal_results_{city_name.replace(' ', '_')}_2024.gpkg")

            # Check if the file already exists in Google Drive
            if os.path.exists(google_drive_save_path):
                print(f"  Results for {city_name} (2024) already exist at '{google_drive_save_path}'. Skipping processing.")
                continue

            try:
                # Load the GeoPackage file
                gdf = gpd.read_file(file_path)
                print(f"  Loaded data for {city_name} (2024).")

                # Ensure the GeoDataFrame has a valid projection for accurate G_Local calculation
                if gdf.crs is None or gdf.crs.is_geographic:
                     print("  Warning: GeoDataFrame has no CRS or is geographic. Reprojecting to EPSG:3035 for accurate G_Local calculation with loaded weights.")
                     gdf_projected_for_g = gdf.to_crs(epsg=3035) # Use a projected CRS
                else:
                     gdf_projected_for_g = gdf

                # Perform G_Local calculations using the function
                g_local_values, g_local_pvalues = calculate_g_local_with_weights(gdf_projected_for_g, wq)

                # Store the results
                if g_local_values is not None and g_local_pvalues is not None:
                     # Add the G_Local values and p-values back to the original gdf (non-projected)
                     gdf['g_local_2024'] = g_local_values
                     gdf['g_local_pvalue_2024'] = g_local_pvalues

                     # Filter for cold spots: p-value < 0.01 and G_Local value < 0
                     # Keep all columns for saving
                     cold_spots_gdf = gdf[(gdf['g_local_pvalue_2024'] < 0.01) & (gdf['g_local_2024'] < 0)].copy()


                     # Save the results to Google Drive as GeoPackage
                     if not cold_spots_gdf.empty:
                         try:
                             cold_spots_gdf.to_file(google_drive_save_path, driver='GPKG')
                             print(f"  Significant cold spots for {city_name} (2024) saved to Google Drive: {google_drive_save_path}")
                         except Exception as save_e:
                            print(f"  Error saving results for {city_name} (2024) to {google_drive_save_path}: {save_e}")

                     else:
                         print(f"  No significant cold spots found for {city_name} (2024) with p < 0.01 and G_Local < 0. No file saved.")


            except Exception as e:
                print(f"  Error processing file {file_path}: {e}")
            finally:
                # Explicitly delete the GeoDataFrame
                if 'gdf' in locals():
                    del gdf
                if 'gdf_projected_for_g' in locals():
                     del gdf_projected_for_g
                if 'cold_spots_gdf' in locals():
                     del cold_spots_gdf
                gc.collect()
                print(f"  Finished G_Local processing for {city_name} (2024). Memory freed.")

else:
    print("\n  Weights matrix not loaded. Skipping G_Local analysis.")


print("\n--- G_Local analysis finished and results saved to Google Drive. ---")
```
